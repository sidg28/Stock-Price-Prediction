```python
adult_data = spark.read.csv("dbfs:/databricks-datasets/adult/adult.data", header=True)
type(adult_data)
adult_data.show()
```

```python
#requires paid version :(
adult_data_rdd = adult_data.rdd
type(adult_data_rdd) 
#will show as spark.rdd.RDD

adult_data_rdd.collect()
#will materialize the data

adult_data_rdd.count()
#will give count

adult_data_rdd.first()
#will give first row


adult_data_rdd.map(lambda row: (row[1], row[3], row[5]))\
    collect()
#apply operation on all rows
#It is a tranformation
#collect method runs after applying transformation 

adult_data_rdd_filtered = adult_data_rdd.filter(lambda row: row[' <=50K'] == ' <=50K')
#filters rows using  column on basis of column name
```

```python
stock_data = spark.read.option('multiline', 'true')\
    .json('/Volumes/workspace/default/playground_data/stock_data.json')
```

```python
display(stock_data)
#all operations can be performed now same as dataframe
```

```python
display(stock_data.select("company_name", "historical_metrics.peak_price", "market_info.currency"))
```

```python
from pyspark.sql import functions as F
display(stock_data.select(F.col("company_name").alias("name"), \
    F.col("historical_metrics").getItem(0).getItem("peak_price").alias("price"), \
    F.col("historical_metrics").getItem(0).getItem("year").alias("year")))
```
