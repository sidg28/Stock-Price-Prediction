```python
insurance_df.crosstab('sex', "smoker")\
    .select('sex_smoker', 'yes', 'no')\
        .display()
#contributes to many columns/partitions so it is a wide tranformation
```

```python
insurance_df.select("age", "sex", "bmi", "region", "charges")\
    .where(insurance_df["age"] > 50)\
        .limit(10)\
            .display()
#filter and where are interchangeable
```

```python
insurance_df.select('age', 'sex', 'bmi', 'region', 'charges')\
    .where(insurance_df['region'].isin(['southwest', 'southeast']))\
    .withColumnRenamed('sex', 'gender')\
    .orderBy('charges')\
    .display()
#renaming of column done, isin is used to specify multiple values
```

```python
spark.sql("""select date,  `close/last`, volume, open
              from aapl_data
              limit 3
          """).display()
```

```python
aapl_subset = spark.sql("""select date,  `close/last`, volume, open
              from aapl_data
              where cast(ltrim('$', high) as float)-cast(ltrim('$', low) as float) > 5
          """)
aapl_subset.createOrReplaceTempView('aapl_subset')
```

```python
spark.sql('select * from aapl_subset').display()
```

```python
spark.sql("select * from aapl_data where cast(ltrim('$', high) as float)-cast(ltrim('$', low) as float) > 5").write.saveAsTable("aapl_high_fluctuations")
```
